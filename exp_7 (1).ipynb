```python
#Reverse a number
num = int(input("Enter a number: "))
rev = 0
while num > 0:
  digit = num % 10
  rev = rev * 10 + digit
  num //= 10
print("Reversed: ",rev)
```

```python
#check pallindrome number/string

num = int(input("Enter a number: "))
temp = num
rev = 0
while num > 0:
  rev = rev*10 + num%10
  num //= 10

if temp == rev:
  print("Pallindrome")
else:
  print("Not Pallindrome")
```

```python
# Check Palindrome using if-else: (works with both numbers and letters)
## numbers and lettes and stored as string
s = "madam"
i, j = 0, len(s)-1
is_pallindrom = True

while i < j:
  if s[i] != s[j]:
    is_pallindrom = False
    break
  i += 1
  j -= 1

if is_pallindrom:
  print("Pallindrome")
else:
  print("Not Pallindrome")
```

```python
#letters
s = input("Enter the string")
i, j = 0, len(s)-1
is_pallindrom = True

while i < j:
  if s[i] != s[j]:
    is_pallindrom = False
    break
  i += 1
  j -= 1

if is_pallindrom:
  print("Pallindrome")
else:
  print("Not Pallindrome")
```

```python
#Reverse Letter
st = input("Enter a string: ")
rev = st[::-1]
if st == rev:
  print("Pallindrome")
else:
  print("Not Pallindrome")
```
